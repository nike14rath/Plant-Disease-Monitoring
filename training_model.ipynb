# Importing Libraries

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

# Data Preprocessing

### Training Image Preprocessing

```python
training_set = tf.keras.utils.image_dataset_from_directory(
    'train',
    labels="inferred",  # directly name will be selected as label here
    label_mode="categorical",  # since the labels are categorical 
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),  # we have chanegd this parameter here
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
)
```

### Valid Image Preprocessing

```python
validation_set = tf.keras.utils.image_dataset_from_directory(
    'valid',
    labels="inferred",  # directly name will be selected as label here
    label_mode="categorical",  # since the labels are categorical 
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(128, 128),  # this  parameter i haven't changeb
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
    
)
```

### Building Model

```python
from tensorflow.keras.layers import Dense, Conv2D, Flatten, MaxPool2D, Dropout
from tensorflow.keras.models import Sequential
```

```python
model = Sequential()
```

```python
## Building Convolutional layer
## Building Convolutional layer
model.add(Conv2D(filters=32,
                 kernel_size = 3,
                 padding = 'same',
                 activation = 'relu',
                 input_shape = [128,128,3]
                )
         )
model.add(Conv2D(filters=32,  # Fixed typo: 'filers' -> 'filters'
                 kernel_size = 3,
                 activation = 'relu',  # Fixed typo: 'activaiton' -> 'activation'
                )
         )
model.add(MaxPool2D(pool_size = 2, strides = 2))  # Fixed typo: 'MaxPoll2D' -> 'MaxPool2D'
```

```python
## Building Convolutional layer
model.add(Conv2D(filters=64,
                 kernel_size = 3,
                 padding = 'same',
                 activation = 'relu',
                )
         )
model.add(Conv2D(filters=64,  # Fixed typo: 'filers' -> 'filters'
                 kernel_size = 3,
            
                 activation = 'relu',  # Fixed typo: 'activaiton' -> 'activation'
                )
         )
model.add(MaxPool2D(pool_size = 2, strides = 2))  # Fixed typo: 'MaxPoll2D' -> 'MaxPool2D'
```

```python
## Building Convolutional layer
model.add(Conv2D(filters=128,
                 kernel_size = 3,
                 padding = 'same',
                 activation = 'relu',

                )
         )
model.add(Conv2D(filters=128,  # Fixed typo: 'filers' -> 'filters'
                 kernel_size = 3,
                 # we are removing the padding here to train our model rapidly (kind of not required)
                 activation = 'relu',  # Fixed typo: 'activaiton' -> 'activation'
                )
         )
model.add(MaxPool2D(pool_size = 2, strides = 2))  # Fixed typo: 'MaxPoll2D' -> 'MaxPool2D'
```

```python
## Building Convolutional layer
model.add(Conv2D(filters=256,
                 kernel_size = 3,
                 padding = 'same',
                 activation = 'relu',

                )
         )
model.add(Conv2D(filters=256,  # Fixed typo: 'filers' -> 'filters'
                 kernel_size = 3,
                 # irrelevent padding here
                 activation = 'relu',  # Fixed typo: 'activaiton' -> 'activation'
                )
         )
model.add(MaxPool2D(pool_size = 2, strides = 2))  # Fixed typo: 'MaxPoll2D' -> 'MaxPool2D'
```

```python
model.add(Conv2D(filters=512,
                 kernel_size = 3,
                 padding = 'same',
                 activation = 'relu',

                )
         )
model.add(Conv2D(filters=512,  # Fixed typo: 'filers' -> 'filters'
                 kernel_size = 3,
                 
                 activation = 'relu',  # Fixed typo: 'activaiton' -> 'activation'
                )
         )
model.add(MaxPool2D(pool_size = 2, strides = 2))
```

```python
model.add(Dropout(0.25)) # to avoid the overfitting 
```

```python
model.add(Flatten())
```

```python
model.add(Dense(units = 1500, activation = 'relu'))
```

```python
model.add(Dropout(0.25))
```

```python
# output layer
model.add(Dense(units = 38, activation = 'softmax', ))
```

### Compiling Model

```python
model.compile(optimizer=tf.keras.optimizers.legacy.Adam(
                learning_rate=0.0001),
                # optimizer = 'adam',
              loss = 'categorical_crossentropy',
              metrics= ['accuracy']
             )
```

```python
model.summary()
```

### Model Training

```python
training_history = model.fit(
    x=training_set,
    validation_data=validation_set,
    epochs=10,
    verbose=1
)
```

#### To avoid Overshooting Loss function
1. Choose small learning rate default 0.001 here we have taken 0.0001
2. There may be chance of underfitting so increase number of neuron
3. Add more Convolutional Layer to extract more feature from images there may be possibilty that model unable to capture relevant feature or model is confusing due to lack of feature so feed with more feature

### Model Evaluation

```python
# model Evaluation of the training set
train_loss, train_acc = model.evaluate(training_set)
```

```python
print(train_loss, train_acc)
```

```python
val_loss, val_acc = model.evaluate(validation_set)
```

```python
print(val_loss, val_acc)
```

```python
y_pred = model.predict(test_set)
y_pred, y_pred.shape
```

```python
# here our task is to select the class with maximum probability 
predicted_categories = tf.argmax(y_pred, axis = 1)
```

```python
predicted_categories
```

```python
true_categoires = tf.concat([y for x,y in test_set], axis = 0)
true_categoires
```

```python
Y_true = tf.argmax(true_categoires, axis = 1)
Y_true
```

```python
from sklearn.metrics import classification_report
```

```python
report = classification_report(
    Y_true,
    predicted_categories,
    labels=range(len(class_name)),
    target_names=class_name
)
print(report)
```
